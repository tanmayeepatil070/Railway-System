## Taking input from user using input() function

```python
number1 = 12
number2 = 32
print(number1 + number2)
print(number1 * number2)
```

```python
number1 = input("Enter number1: ")
number2 = input("Enter number2: ")
print(number1 + number2)
print(number1 * number2)
```

```python
num1 = int(input("Enter num1: "))
print(type(num1))
num2 = int(input("Enter num2: "))
print(type(num2))
print(num1 + num2)
print(num1 * num2)
      
```

```python
num1 = float(input("Enter num1: "))
print(type(num1))
num2 = float(input("Enter num2: "))
print(type(num2))
print(num1 + num2)
print(num1 * num2)
# in float() we can add int() value but it will take as 0.0 behind the scenes
```

# Control Flow in Python

## Conditional Statements

### Only If Statements

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote")
    print("Thank You")

print("Successfully logged out")
```

```python
PersonAge = int(input("Enter you age: "))
City = input("Enter you city: ")

if PersonAge >= 18:
    if City == "Nagpur":
        print("You are eligible to vote")
        print("Thank You")

print("Succcessfully logged out")
```

### if-else statement

```python
marks = int(input("Enter your marks: "))

if marks >= 40:
    print("Congratulations😂")
    print("You have passed your exam")
else:
    print("You are failed")
```

### nested if-else statement

```python
n1 = int(input("Enter your num1 "))
n2 = int(input("Enter your num2 "))

if n1 >= n2:
    if n1 == n2:
        print(f'{n1}is equal to {n2}')
    else : 
        print(f'{n1} is greater than {n2}')
else : 
    print(f'{n1} is less than {n2}')
```
